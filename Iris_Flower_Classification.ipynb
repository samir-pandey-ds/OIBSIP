# Iris Flower Classification

## 1. Project Objective
The objective of this project is to predict the species of an iris flower using its physical measurements.

## 2. Dataset
We will use the built-in Iris dataset provided by Scikit-learn.

## 3. Technologies Used
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Scikit-learn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
```

```python
iris = load_iris()

print("Dataset successfully loaded!")
print("Dataset type:", type(iris))
```

```python
print("Feature names:")
print(iris.feature_names)

print("\nTarget names:")
print(iris.target_names)

print("\nDataset shape:")
print(iris.data.shape)

print("\nTarget shape:")
print(iris.target.shape)
```

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    categories=iris.target_names
)

df.head()
```
